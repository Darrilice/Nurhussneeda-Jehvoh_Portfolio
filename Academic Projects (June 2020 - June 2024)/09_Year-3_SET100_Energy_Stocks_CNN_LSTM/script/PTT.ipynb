2. เตรียมข้อมูลก่อนเข้าโมเดล — time series data format

```python
from matplotlib import pyplot
import pandas as pd
from IPython.core.display import display , HTML
```

```python
display(HTML("<style>.container { width:90% !important; }</style>"))
pd.options.display.max_columns = 999
pd.options.display.max_rows = 999
```

```python
# function สำหรับ preprocess ข้อมูล time series หลายตัวแปร
def series_to_supervised(data , n_in = 1 , n_out = 1 , dropnan = True , feat_name = None) :
    
    n_vars = 1 if type(data) is list else data.shape[1]
    df = pd.DataFrame(data)
    cols , names = list() , list()
    
    # input sequence (t-n, ... t-1)
    for i in range(n_in , 0 , -1) :
        cols.append(df.shift(i))
        names += [f'{feat_name[j]}(t-{i})' for j in range(n_vars)]
    
    # forecast sequence (t, t+1, ... t+n)
    for i in range(0 , n_out) :
        cols.append(df.shift(-i))
        if i == 0 :
            names += [f'{feat_name[j]}(t)' for j in range(n_vars)]
        else :
            names += [f'{feat_name[j]}(t+{i})' for j in range(n_vars)]
    
    # put it all together
    agg = pd.concat(cols , axis = 1)
    agg.columns = names
    
    # drop rows with NaN values
    if dropnan :
        agg.dropna(inplace = True)
    return agg
```

```python
dataset = ptt_price
values = dataset.values
values = values.astype('float32')
```

```python
# ทำ scaling ข้อมูลด้วยวิธี min max scaling เปลี่ยน scale ข้อมูลแต่ละ column ให้อยู่ระหว่าง [0,1] และเก็บค่า min max แต่ละ column ไว้สำหรับทำ rescale ข้อมูลภายหลัง
min_dict = dict()
max_dict = dict()
for col in dataset.columns :
  min_dict[col] = dataset[col].min()
  max_dict[col] = dataset[col].max()
  dataset[col] = (dataset[col] - dataset[col].min()) / (dataset[col].max() - dataset[col].min())
```

3. เตรียมข้อมูลก่อนเข้าโมเดล — train&test split

```python
# ทำ train & test split
values = reframed.values
n_train_percent = 0.8
split = int(values.shape[0] * 0.8)
train = values[:split , :]
test = values[split: , :]
```

```python
# เก็บค่า datetime ของ train & test
date_reframed = dataset.index[reframed.index]
date_train = date_reframed[: split]
date_test = date_reframed[split :]
```

```python
# split feature และ target
train_X , train_y = train[: , : -30] , train[: , -30 :]
test_X , test_y = test[: , : -30] , test[: , -30 :]
```

```python
len(train_X)
```

```python
# reshape input ให้เป็น 3D (samples, timesteps, features)
train_X = train_X.reshape((train_X.shape[0] , 1 , train_X.shape[1]))
test_X = test_X.reshape((test_X.shape[0] , 1 , test_X.shape[1]))
```

4. สร้างโมเดล LSTM และ train โมเดล

```python
from keras.models import Sequential
from keras.layers import Dense
from keras.layers import LSTM
```

```python
model = Sequential()
model.add(LSTM(200 , 
               input_shape = (train_X.shape[1] ,  train_X.shape[2]) , 
               activation = "relu"))
```

```python
model.add(Dense(30))
model.compile(loss = 'mae' , 
              optimizer = 'adam' , 
              metrics=['accuracy'])
```

```python
history = model.fit(train_X , 
                    train_y , 
                    epochs = 500 , 
                    batch_size = 72 , 
                    validation_data = (test_X , test_y) , 
                    verbose = 2 , 
                    shuffle = False)
```

```python
import plotly
import plotly.graph_objs as go
```

```python
h1 = go.Scatter(y = history.history['loss'] , 
                    mode = "lines" ,
                    line = dict(
                        width = 2 ,
                        color = 'blue') ,
                        name = "loss"
                   )
h2 = go.Scatter(y = history.history['val_loss'] , 
                    mode = "lines" ,
                    line = dict(
                        width = 2 ,
                        color = 'red') ,
                        name = "val_loss"
                   )
                   
data = [h1 , h2]
layout1 = go.Layout(title = 'Loss' ,
                   xaxis=dict(title = 'epochs') ,
                   yaxis=dict(title = ''))
fig1 = go.Figure(data , layout = layout1)
plotly.offline.iplot(fig1)
```

5. สรุปผล

```python
# ทำ prediction
yhat = model.predict(test_X)
test_X = test_X.reshape((test_X.shape[0] , test_X.shape[2]))
```

```python
# rescale ข้อมูลกลับ
inv_yhat = yhat * (max_dict['adjclose'] - min_dict['adjclose']) + min_dict['adjclose']
inv_y = test_y * (max_dict['adjclose'] - min_dict['adjclose']) + min_dict['adjclose']
```

```python
# พล็อตราคาจริงของวันที่ทำ prediction ล่วงหน้า 30 วัน + ราคาย้อนหลัง 365 วัน (สีน้ำเงิน) และ ราคา forecast (สีแดง)
import numpy as np

pyplot.figure(figsize = (15 , 8))

idx = np.random.randint(0 , len(yhat - 30))

if idx < 365 :
  date_actual = date_test[: idx + 30]
else :
  date_actual = date_test[idx - 365 : idx + 30]
_ = pyplot.plot(date_actual , inv_y[max(0 , idx - 365) : idx + 30 , 0] ,'b')
_ = pyplot.plot(date_test[idx : idx + 30] , inv_yhat[idx] ,'r')

pyplot.title('Adjust Close')
pyplot.xlabel('Date')
```

```python
from sklearn.metrics import mean_squared_error
from math import sqrt
```

```python
mse = mean_squared_error(inv_y , inv_yhat)

print('Mean Square Error : ' , mse)
```

```python
rmse = sqrt(mse)

print('Root Mean Square Error : ' , rmse)
```
