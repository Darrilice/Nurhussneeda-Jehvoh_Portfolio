```python
from bs4 import BeautifulSoup
import pandas as pd
import requests
import re
```

```python
url ='https://xn--12cas3c2av3m3a0g7c.com/%E0%B8%9D%E0%B8%B6%E0%B8%81%E0%B8%87%E0%B8%B2%E0%B8%99/%E0%B8%97%E0%B8%B1%E0%B9%89%E0%B8%87%E0%B8%AB%E0%B8%A1%E0%B8%94?page=2'
response = requests.get(url)
```

```python
response
```

```python
page_html = BeautifulSoup(response.text, 'html.parser')
```

```python
train = page_html.find_all("div" , class_="job-post-box")
name = []
company = []
location = []
number = []
picture = []
```

```python
print(train)
```

```python
for container in train :
  names = container.h6.get_text(strip = True)
  name.append(names)

  companies = container.find('div' , 'job-post-box--company').a.get_text(strip = True)
  company.append(companies)

  locations = container.find('div' , 'job-post-box--location').a.get_text(strip = True)
  location.append(locations)

  numbers = container.find('div' , 'job-post-box--capacity no-wrap').get_text(strip = True)
  number.append(numbers)

  pictures = container.find('img' , 'job-post-box__logo')
  pictures = container.img['src']
  picture.append(pictures)
```

```python
df = pd.DataFrame({
  'JOB DESCRIPTION': name ,
  'COMPANY': company ,
  'LOCATION': location ,
  'EMPLOYMENT RATE': number ,
  'PICTURE': picture,
})
```
