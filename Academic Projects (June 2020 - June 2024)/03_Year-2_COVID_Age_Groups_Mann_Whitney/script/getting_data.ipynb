```python
df = pd.read_csv("coviddd.csv")
print(df )
```

```python
condition_1 = df['age'] > 11
condition_2 = df['age'] < 21
teen = df[ (condition_1) & (condition_2)]

print(teen)
```

```python
condition_1 = df['age'] > 60
condition_2 = df['age'] < 81
older = df[ (condition_1) & (condition_2)]

print(older)
```

```python
older.to_csv("Age-Older.csv", encoding = 'utf-8',index= False)
```
