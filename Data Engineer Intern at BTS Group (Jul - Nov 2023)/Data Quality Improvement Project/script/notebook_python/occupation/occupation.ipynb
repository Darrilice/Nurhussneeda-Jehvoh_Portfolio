```python
# make the data to lower case
# remove . and ,
# replace - to white space
def standard_transform(text_input):
    text_input = text_input.lower().replace('.' , '').replace('-' , ' ').replace(',' , '')
    return text_input



# create column for transform data
data_occu['occupation_transform'] = data_occu['occupation'].apply(lambda occupation : standard_transform(occupation))
```

```python
# define type of occupations are 'business owner' , 'employee' , 'officer' , 'other'
def typ(input_occu) :
    if re.search(r'^(b|ow.*|ow.*[0-9]\s?[a-z]|เ.*จด|ธ.*จด|เ.*[0-9]|ธ.*[0-9])+' , input_occu) :
        return 'business owner'
    elif re.search(r'^(em|ล)' , input_occu) :
        return 'employee'
    elif re.search(r'^(of|พ)' , input_occu) :
        return 'officer'
    else :
        return 'other'
    
    
    
# create column for define type
data_occu['type'] = data_occu['occupation_transform'].apply(lambda occupation_transform : typ(occupation_transform))
```

```python
# choose only english letters
def take_eng(input_occu) :
    try :
        if re.search(r'[a-zA-Z]+' , input_occu) :
            return input_occu
    except AttributeError :
        return None
    return None



# create column for english letters
data_occu['df_eng'] = data_occu['occupation_transform'].apply(lambda occupation_transform : take_eng(occupation_transform))
```

```python
# choose only thai letters
def take_thai(input_occu) :
    try :
        if re.search(r'[ก-๛]+' , input_occu) :
            return input_occu
    except AttributeError :
        return None
    
    
    
# create column for thai letters
data_occu['df_thai'] = data_occu['occupation_transform'].apply(lambda occupation_transform : take_thai(occupation_transform))
```

```python
# create data frame for case = eng and every types
occu_eng_b = data_occu[data_occu['df_eng'].notnull()][data_occu['type'] == 'business owner']
occu_eng_e = data_occu[data_occu['df_eng'].notnull()][data_occu['type'] == 'employee']
occu_eng_o = data_occu[data_occu['df_eng'].notnull()][data_occu['type'] == 'officer']
occu_eng_ot = data_occu[data_occu['df_eng'].notnull()][data_occu['type'] == 'other']

# create data frame for case = tha and every types
occu_thai_b = data_occu[data_occu['df_thai'].notnull()][data_occu['type'] == 'business owner']
occu_thai_e = data_occu[data_occu['df_thai'].notnull()][data_occu['type'] == 'employee']
occu_thai_o = data_occu[data_occu['df_thai'].notnull()][data_occu['type'] == 'officer']
occu_thai_ot = data_occu[data_occu['df_thai'].notnull()][data_occu['type'] == 'other']

# del original data frame
del data_occu
```

```python
# create standard format
def std_occu(text_input,lang,typ) :
    for pattern,value in pattern_dict[lang][typ].items():
        if re.search(pattern , text_input):
            return value
        else:
            continue
    return None



# define list of english and thai data frames
# define the order of type
eng_df_list = [occu_eng_b,occu_eng_e,occu_eng_o,occu_eng_ot]
thai_df_list = [occu_thai_b,occu_thai_e,occu_thai_o,occu_thai_ot]
key_list = ['business owner', 'employee', 'officer', 'other']

# create new column of case = eng and case = tha for standardize format
for eng_df,key in zip(eng_df_list,key_list):
    eng_df['standard_occupation'] = eng_df['df_eng'].apply(lambda df_eng : std_occu(df_eng , 'eng' , key))
    
for thai_df,key in zip(thai_df_list,key_list):
    thai_df['standard_occupation'] = thai_df['df_thai'].apply(lambda df_thai : std_occu(df_thai , 'tha' , key))
```

```python
# combined data frame of english , type and thai , type
occupation = pd.concat([occu_eng_b,occu_eng_e,occu_eng_o,occu_eng_ot,occu_thai_b,occu_thai_e,occu_thai_o,occu_thai_ot])
```

```python
# last result
def result(text_input, standard):
    if isinstance(text_input , str) and standard == None :
        return text_input , 0
    
    elif text_input == standard :
        return text_input , 1
    
    elif text_input != standard :
        return standard , 2
    
    elif text_input == None and standard == None :
        return text_input , 3
    
    
    
    
occupation[['std_output' , 'status']] = occupation.apply(lambda row: pd.Series(result(row['occupation'], row['standard_occupation'])), axis=1)
```

```python
selected_cols = ['id','occupation','standard_occupation','std_output','status']

occupation = occupation[selected_cols]
```
